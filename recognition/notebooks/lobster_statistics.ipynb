```python
from dataset import LOBSTERLevel10Dataset
import numpy as np
import pandas as pd
```

```python
books = LOBSTERLevel10Dataset(train=True).books
```

```python
midpoints = 0.5 * (books['ASKp1'] + books['BIDp1'])
midpoints = np.log(midpoints)
np.mean(midpoints), np.std(midpoints)
```

```python
spreads = books['ASKp1'] - books['BIDp1']
spreads = np.log(spreads)
np.mean(spreads), np.std(spreads)
```
