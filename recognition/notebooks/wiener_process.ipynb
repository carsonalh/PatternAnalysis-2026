# Wiener Process Simulation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)

SAMPLES = 500
T = 1.0

dt = T / SAMPLES

t = np.linspace(0.0, T, SAMPLES)
y = np.concatenate(
    (
        np.array([0]),
        np.cumsum(dt * np.random.normal(size=SAMPLES))[:SAMPLES-1]
    ),
    axis=0
)

plt.plot(t, y)
plt.xlabel("t")
plt.ylabel("y")
plt.grid()
```
